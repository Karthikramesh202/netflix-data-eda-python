```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Load the Netflix dataset from a public source
url = "https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/netflix_titles.csv"
df = pd.read_csv(url)

# Display the first 5 rows to verify it loaded correctly
df.head()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Fill missing values in text columns with 'Unknown'
df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('Unknown', inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['rating'].fillna('Not Rated', inplace=True)

# Drop rows with missing date_added or rating if any remain
df.dropna(subset=['date_added'], inplace=True)

# Verify that missing values are now zero
df.isnull().sum()
```

```python
# Set the style for the plot
sns.set_theme(style="darkgrid")

# Create a count plot for Content Type (Movie vs TV Show)
plt.figure(figsize=(6, 4))
sns.countplot(x='type', data=df, palette='Set2')

# Add title and labels
plt.title('Count of Movies vs TV Shows on Netflix', fontsize=12)
plt.xlabel('Content Type', fontsize=10)
plt.ylabel('Count', fontsize=10)

# Show the plot
plt.show()
```
